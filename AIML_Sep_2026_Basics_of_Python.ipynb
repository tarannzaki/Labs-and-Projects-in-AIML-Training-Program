```python
## Sets
'''a Set is an unordered collection of data types that is iterable, mutable and has no duplicate elements.
The order of elements in a set is undefined though it may consist of various elements.
The major advantage of using a set, as opposed to a list,
 is that it has a highly optimized method for checking whether a specific element is contained in the set.
'''
set_unique = set([1,1,2,3,5,6,'Lab1'])
print(set_unique) ##notice it is unordered
last_el = set_unique.pop()
set_unique.add((1,2))


print(last_el)
print(set_unique)




```

```python
# 1. STRING METHODS
text = "Hello World"

print(text.upper())       # Converts to uppercase
print(text.lower())       # Converts to lowercase
print(text.replace("World", "Python"))  # Replaces text
print(text.split())       # Splits string into a list
print(text.find("World")) # Finds position of substring


# 2. LIST METHODS
my_list = [1, 2, 3]

my_list.append(4)         # Adds an element
my_list.extend([5, 6])    # Adds multiple elements
my_list.reverse()         # Reverses the list
my_list.sort()            # Sorts the list
my_list.pop()             # Removes the last element

print(my_list)


# 3. TUPLE METHODS
my_tuple = (1, 2, 2, 3, 4)

print(my_tuple.count(2))  # Counts occurrences
print(my_tuple.index(3))  # Finds index of an element


# 4. SET METHODS
my_set = {1, 2, 3}

my_set.add(4)             # Adds an element
my_set.update([5, 6])     # Adds multiple elements
my_set.remove(2)          # Removes an element
print(my_set)


# 5. DICTIONARY METHODS
my_dict = {"name": "Taran", "age": 19}

print(my_dict.keys())     # Returns all keys
print(my_dict.values())   # Returns all values
print(my_dict.items())    # Returns key-value pairs

my_dict.update({"age": 20})  # Updates dictionary
my_dict.pop("age")           # Removes a key
print(my_dict)
```

```python
def check_even_list(num_list):

    even_numbers = []

    # Go through each number
    for number in num_list:
        # Once we get a "hit" on an even number, we append the even number
        if number % 2 == 0:
            even_numbers.append(number)
        # Don't do anything if its not even
        else:
            pass
    # Notice the indentation! This ensures we run through the entire for loop
    return even_numbers
```

##### Given a list having Names, work_hours, and gender, Write a function to print name of the female worker that worked the most hours. Also how much do should she be paid if the pay is $ 20 per hour.

##### work_hours = [('Abby',100 , 'F'),('Billy',400, 'M'),('Cassie',800,'F'), ('Maggi',600,'F'),('Alex',500,'M'),('Raj',225,'M'),('Penny',920,'F'),('Ben',300,'M')]

##### Answer : the female worker that worked the most hours is Penny and she should be paid 18400

```python
#Question 1
def reverse_and_add(A, B):
    reversed_A = A[::-1]
    return reversed_A + B


A = "Hello"
B = "World"

result = reverse_and_add(A, B)

print(result)


#Question 2
def highest_paid_female(work_hours):
    max_hours = 0
    female_worker = ""

    for name, hours, gender in work_hours:
        if gender == 'F' and hours > max_hours:
            max_hours = hours
            female_worker = name

    salary = max_hours * 20

    print("Female worker:", female_worker)
    print("Hours worked:", max_hours)
    print("Salary: $", salary)


work_hours = [
    ('Abby', 100, 'F'),
    ('Billy', 400, 'M'),
    ('Cassie', 800, 'F'),
    ('Maggi', 600, 'F'),
    ('Alex', 500, 'M'),
    ('Raj', 225, 'M'),
    ('Penny', 920, 'F'),
    ('Ben', 300, 'M')
]

highest_paid_female(work_hours)
```

```python
import numpy as np
```

```python

a = np.array([1,1,2,3,4,5,5,6,1]) ## np.array converts given list to array

b = a>1 ## important comparison operation, where frequently used in manipulation and image processing.

print(b)
print(a[b]) ## [printing only those values in a which are greater than 1]
```

```python
a_range = np.arange(10,19).reshape(3,3) ## create a 3x3 array with values in range 10-19
a_range
```

```python
## Indexing in arrays works same as that of list

a_range[0] # printing all the columns of first row
```

```python
a_range[:,2] #printing all the rows of second column
```

```python
iden = np.eye(3) #idnetity matrix of given size
iden
```

```python
## adding two matrices
summed = a_range + iden
summed
```

```python
### arrays support normal matrix multiplication that you are used to, point-wise multiplication
### and dot product as well.

mul = a_range@iden ## normal multiplication
mul
```

```python
## point wise multiplication
p_mul = a_range * iden
p_mul
```

```python
## Transpose of a matrix.

mtx_t = mul.T
mtx_t
```

```python
### Here we are changing the values of last row of the transposed matrix.
### basically point wise multiplying the values of last row with 1,2 and 3

mtx_t[2] = mtx_t[2]*[1,2,3] ## indexing, point wise multiplication and mutation of values
mtx_t
```

```python
## Just like the greater than 1 (a>1) example we saw earlier.
## here we are checking if the elements are divisible by 2 (%), and if they are, then replace by 0.

mtx_t[(mtx_t % 2 == 0)] = 0 ## convert even elements of the matrix to zero.
mtx_t
```

## Creating a dataframe.
#### Task: Study about other methods of creating dataframe (for example: using Pandas Series, Lists etc.)

```python
import pandas as pd

# Creating DataFrame using a dictionary
data = {
    'Name': ['Taran', 'Rahul', 'Arjun'],
    'Age': [19, 20, 21],
    'Marks': [90, 85, 88]
}

df1 = pd.DataFrame(data)
print("DataFrame using Dictionary:")
print(df1)


# Creating DataFrame using a list of lists
data_list = [
    ['Taran', 19, 90],
    ['Rahul', 20, 85],
    ['Arjun', 21, 88]
]

df2 = pd.DataFrame(data_list, columns=['Name', 'Age', 'Marks'])

print("\nDataFrame using List:")
print(df2)


# Creating DataFrame using Pandas Series
name = pd.Series(['Taran', 'Rahul', 'Arjun'])
age = pd.Series([19, 20, 21])
marks = pd.Series([90, 85, 88])

df3 = pd.DataFrame({
    'Name': name,
    'Age': age,
    'Marks': marks
})

print("\nDataFrame using Pandas Series:")
print(df3)
```
